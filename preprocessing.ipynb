# Handling Missing Values

```python
df.isna().sum()
```

3 cols have missing values : Gross and Meta Score are important ones

```python
df['Certificate'] = df['Certificate'].fillna("NA")
```

Fill most frequent value in missing place of Gross and Meta Score

```python
freqMeta=df['Meta_score'].mode()[0]
df['Meta_score']=df['Meta_score'].fillna(freqMeta)
```

```python
freqGross=df['Gross'].mode()[0]
df['Gross']=df['Gross'].fillna(freqGross)
```

```python
print(df['Certificate'].isna().sum())
print(df['Meta_score'].isna().sum())
print(df['Gross'].isna().sum())
```

No missing Values remain

# Examining Values in Columns

```python
df.info()
```

```python
print('Released_Year:\n',df['Released_Year'].unique())
print('Certificate:\n',df['Certificate'].unique())
```

Removing PG from Released_Year :

```python
df=df[df['Released_Year']!="PG"]
```

```python
df['Released_Year'].unique()
```

Removing min from Runtime:

```python
df['Runtime']=df['Runtime'].str.replace('min','')
```

```python
df.head()
```

Removing Commas from Gross:

```python
df['Gross']=df['Gross'].str.replace(',','')

df.head()
```

Converting DataTypes:

```python
df['Released_Year'] = df['Released_Year'].astype(int)
df['Runtime'] = df['Runtime'].astype(int)
df['Gross'] = df['Gross'].astype(np.int64)
df['Meta_score'] = df['Meta_score'].astype(int)
```
